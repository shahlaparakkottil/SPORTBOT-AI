```python
import pandas as pd

# Load CSV
df = pd.read_csv(r"C:\projects\sports_chatbot_2026_dataset.csv")

df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# Convert text to vectors
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(df["question"])

# Target column
y = df["intent"]

# Train model
model = LogisticRegression()
model.fit(X, y)

print("Model Trained Successfully ✅")
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def chatbot(user_input):

    # Convert user text to vector
    user_vec = tfidf.transform([user_input])

    # Predict intent
    predicted_intent = model.predict(user_vec)[0]

    # Filter dataset by predicted intent
    filtered_df = df[df["intent"] == predicted_intent]

    # Vectorize filtered questions
    filtered_vecs = tfidf.transform(filtered_df["question"])

    # Compute similarity
    similarity = cosine_similarity(user_vec, filtered_vecs)

    # Get best matching question
    best_index = np.argmax(similarity)

    # Return best response
    return filtered_df.iloc[best_index]["response"]
```

```python
print("Sports Chatbot Ready! Type 'exit' to stop.\n")

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Bot: Goodbye 👋")
        break

    response = chatbot(user_input)
    print("Bot:", response)
```
